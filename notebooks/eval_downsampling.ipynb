# Downsampling

```python
names = [
    "MinMaxDownsampler", 
    "M4Downsampler", 
    "MinMaxLTTBDownsampler", 
    "LTTBDownsampler"
]

for i, method in enumerate([
    MinMaxDownsampler(), 
    M4Downsampler(), 
    MinMaxLTTBDownsampler(), 
    LTTBDownsampler()]):
    for sampling_factor in [2, 3, 4, 5]:
        print("------------------------")
        print("Sampling factor: ", sampling_factor)
    
        start = time.time()
    
        if sampling_factor > 1:        
            max_points = n * 2 // sampling_factor
            index = method.downsample(ts, n_out=max_points)        
            print(index.shape, ts.shape)
            ts_raw = ts.iloc[index]
        else:
            ts_raw = ts
    
        ll = l // sampling_factor
        mm = Motiflets(ds_name, ts_raw, backend="scalable", n_jobs=4)
        
        mm.fit_k_elbow(
            k_max, ll, 
            plot_elbows=False,
            plot_motifs_as_grid=False
        )
        duration = time.time() - start
        memory_usage = mm.memory_usage
    
        new_filename = f"images/penguin_downsample_{names[i]}_{sampling_factor}.pdf"
        mm.plot_motifset(new_filename)
            
        current = [n, sampling_factor, duration, memory_usage]
        
        print("\tDiscovered motiflets in", duration, "seconds")
        print("\t", current)
```
